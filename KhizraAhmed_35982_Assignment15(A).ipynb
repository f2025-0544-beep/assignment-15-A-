```python
df=pd.read_csv('ride_hailing_dataset.csv')
```

Q1: What is the average fare by service type?

```python
df.groupby('service_type')['fare'].mean().round(2)
```

Q2: Which city has the most trips?

```python
most_trip=df['city'].value_counts()
print(most_trip)
if most_trip['Karachi'] > most_trip['Lahore'] and most_trip['Karachi'] > most_trip['Islamabad'] and most_trip['Karachi'] > most_trip['Rawalpindi'] :
    print("Karachi has the highest number of trips")
elif most_trip['Lahore'] > most_trip['Karachi'] and most_trip['Lahore'] > most_trip['Islamabad']and most_trip['Lahore'] > most_trip['Rawalpindi']:
    print("Lahore has the highest number of trips")
elif most_trip['Islamabad'] > most_trip['Karachi'] and most_trip['Islamabad'] > most_trip['Lahore']and most_trip['Islamabad'] >most_trip['Rawalpindi']:
    print("Islamabad has the highest number of trips")
else:
    print("Rawalpindi has the highest number of trips")
```

Q3: What is the average distance per trip?

```python
avg_distance=df['distance_km'].mean().round(2)
print('The average distance per trip is',avg_distance)
```

Q4: Compare trip ratings by gender

```python
df.groupby('gender')['trip_rating'].mean().round(2)
```

Q5: Make a bar chart of service type vs. average fare.

```python
import matplotlib.pyplot as plt
```

```python
avg_fare=df.groupby('service_type')['fare'].mean()
```

```python
categories = avg_fare.index
values = avg_fare.values
plt.bar(categories, values, color='violet', width=0.5)
plt.xlabel('Service Type')
plt.ylabel('Average Fare')
plt.title('Average Fare by Service Type')
plt.show()
```
